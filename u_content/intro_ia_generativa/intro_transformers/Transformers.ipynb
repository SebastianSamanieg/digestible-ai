## Modelos de NLP

Hay algunos pasos que son parte del proceso de interpretación dentro de la función pipeline y sus modelos:

1- El texto es pre-procesado en un formato que el modelo pueda entender\
2- Los textos pre-procesados entran al modelo\
3- La prediction del modelo es post-procesada, así tu puedes entender el resultado.

### Zero-shot classification

```python
classifier = pipeline("zero-shot-classification")
classifier(
    "This is a course about the Transformers library",
    candidate_labels=["education", "politics", "business"],
)
```

```python
classifier = pipeline("zero-shot-classification", model="tasksource/deberta-small-long-nli")
classifier(
    "This is a course about the Transformers library",
    candidate_labels=["education", "politics", "business"],
)
```

Este pipeline es llamada zero-shot porque no necesitas hacer ningun fine-tunning del modelo sobre tus datos para usarlo. El puede calificar directamente la lista de etiquetas que tu desees evaluar.


```python
from transformers import pipeline
classifier = pipeline("zero-shot-classification", model="Xenova/nli-deberta-v3-xsmall")

sequence_to_classify = "Tribunal de Caldas negó demanda contra Presidencia por transmisión de consejo de ministros"
candidate_labels = ["politica", "economia", "entretenimiento", "ambiente"]
output = classifier(sequence_to_classify, candidate_labels, multi_label=False)
print(output)
```

### Preprocesamiento con un tokenizador
Al igual que otras redes neuronales, los modelos **Transformer** no pueden procesar texto sin procesar directamente, por lo que el primer paso de nuestra canalización es convertir las entradas de texto en números que el modelo pueda entender. Para ello utilizamos un tokenizador, que se encargará de:

1.   Dividir la entrada en palabras, subpalabras o símbolos (como signos de puntuación) que se denominan tokens.
2.   Asignación de cada token a un número entero.
3.   Adición de entradas adicionales que pueden ser útiles para el modelo.

Todo este preprocesamiento debe realizarse exactamente de la misma manera que cuando se entrenó previamente el modelo, por lo que primero debemos descargar esa información del Model Hub. Para ello, utilizamos la clase AutoTokenizer y su método from_pretrained(). Usando el nombre del punto de control de nuestro modelo, recuperará automáticamente los datos asociados con el tokenizador del modelo y los almacenará en caché (por lo que solo se descarga la primera vez que ejecuta el código a continuación).

```python
from transformers import AutoTokenizer

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
```

```python
tokenizer
```

Una vez que tengamos el tokenizador, podemos pasarle directamente nuestras oraciones y obtendremos un diccionario que está listo para alimentar a nuestro modelo. Lo único que queda por hacer es convertir la lista de ID de entrada en tensores.

Puede usar 🤗 Transformers sin tener que preocuparse por qué marco de ML se usa como backend; puede ser PyTorch o TensorFlow, o Flax para algunos modelos. Sin embargo, los modelos Transformer solo aceptan tensores como entrada.

```python
raw_inputs = [
    "I've been waiting for a Machine Learning course my whole life.",
    "I hate this so much!",
]
inputs = tokenizer(raw_inputs, padding=True, truncation=True, return_tensors="pt")
print(inputs)
```

La salida en sí es un diccionario que contiene dos claves, input_ids y attention_mask. input_ids contiene dos filas de enteros (uno para cada frase) que son los identificadores únicos de los tokens de cada frase.

Podemos descargar nuestro modelo preentrenado de la misma manera que lo hicimos con nuestro tokenizador. 🤗 Transformers proporciona una clase AutoModel que también tiene un método from_pretrained():

```python
from transformers import AutoModel

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
model = AutoModel.from_pretrained(checkpoint)
```

```python
model
```

Esta arquitectura contiene solo el módulo base Transformer: dadas algunas entradas, emite lo que llamaremos estados ocultos, también conocidos como características. Para cada entrada del modelo, recuperaremos un vector de alta dimensión que representa la comprensión contextual de esa entrada por parte del modelo Transformer.

La salida vectorial del módulo Transformer suele ser grande. Generalmente tiene tres dimensiones:

1. Tamaño de lote: El número de secuencias procesadas a la vez (2 en nuestro ejemplo).
2. Longitud de la secuencia: La longitud de la representación numérica de la secuencia (16 en nuestro ejemplo).
3. Tamaño oculto: la dimensión vectorial de cada entrada del modelo.

Se dice que es de "alta dimensión" debido al último valor. El tamaño oculto puede ser muy grande (768 es común para los modelos más pequeños, y en los modelos más grandes puede llegar a 3072 o más).

```python
outputs = model(**inputs)
print(outputs.last_hidden_state.shape)
```

Hay muchas arquitecturas diferentes disponibles en 🤗 Transformers, cada una diseñada para abordar una tarea específica. He aquí una lista no exhaustiva:

*  Model (retrieve the hidden states)
*  ForCausalLM
*  ForMaskedLM
*  ForMultipleChoice
*  ForQuestionAnswering
*  ForSequenceClassification
*  ForTokenClassification



Para nuestro ejemplo, necesitaremos un modelo con un encabezado de clasificación de secuencia (para poder clasificar las oraciones como positivas o negativas). Por lo tanto, en realidad no usaremos la clase AutoModel, sino AutoModelForSequenceClassification:

```python
from transformers import AutoModelForSequenceClassification

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
model = AutoModelForSequenceClassification.from_pretrained(checkpoint)
outputs = model(**inputs)
```

```python
outputs
```

Ahora, si nos fijamos en la forma de nuestras salidas, la dimensionalidad será mucho menor: la cabeza del modelo toma como entrada los vectores de alta dimensión que vimos antes, y genera vectores que contienen dos valores (uno por etiqueta):

```python
print(outputs.logits.shape)
```

Dado que tenemos solo dos oraciones y dos etiquetas, el resultado que obtenemos de nuestro modelo es de forma 2 x 2.

**Proecesando la salida del modelo**
Los valores que obtenemos como resultado de nuestro modelo no necesariamente tienen sentido por sí mismos. Echemos un vistazo:

```python
print(outputs.logits)
```

 No se trata de probabilidades, sino de logits, las puntuaciones brutas y no normalizadas generadas por la última capa del modelo. Para convertirse en probabilidades, deben pasar por una capa SoftMax (todos los 🤗 modelos de Transformers emiten los logits, ya que la función de pérdida para el entrenamiento generalmente fusionará la última función de activación, como SoftMax, con la función de pérdida real, como la entropía cruzada):

```python
import torch

predictions = torch.nn.functional.softmax(outputs.logits, dim=-1)
print(predictions)
```

 Para obtener las etiquetas correspondientes a cada posición, podemos inspeccionar el atributo id2label de la configuración del modelo:

```python
model.config.id2label
```

#### Encoding

 La codificación se realiza en un proceso de dos pasos: la tokenización, seguida de la conversión a ID de entrada.

 Como hemos visto, el primer paso es dividir el texto en palabras (o partes de palabras, símbolos de puntuación, etc.), normalmente llamadas tokens. Hay varias reglas que pueden gobernar ese proceso, por lo que necesitamos crear una instancia del tokenizador con el nombre del modelo, para asegurarnos de que usamos las mismas reglas que se usaron cuando se entrenó previamente el modelo.

El segundo paso es convertir esos tokens en números, para que podamos construir un tensor a partir de ellos y alimentarlos al modelo. Para ello, el tokenizador cuenta con un vocabulario, que es la parte que descargamos cuando lo instanciamos con el método from_pretrained(). De nuevo, tenemos que usar el mismo vocabulario que se utilizó cuando se entrenó previamente el modelo.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

sequence = "Using a Transformer network is simple"
tokens = tokenizer.tokenize(sequence)

print(tokens)
```

Obtener los ids

```python
ids = tokenizer.convert_tokens_to_ids(tokens)

print(ids)
```

**Decodificado**

```python
decoded_string = tokenizer.decode([7993, 170, 11303, 1200, 2443, 1110, 3014])
print(decoded_string)
```
