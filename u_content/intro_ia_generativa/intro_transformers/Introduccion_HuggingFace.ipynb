# Introducción a Hugging Face y Aplicaciones en NLP

## 1. ¿Qué es Hugging Face?
Hugging Face es una plataforma líder en NLP que ofrece modelos preentrenados, datasets y herramientas útiles para construir aplicaciones de lenguaje natural como análisis de sentimientos, generación de texto, clasificación, etc.

## 2. Aplicaciones Comunes de Hugging Face
- Análisis de sentimientos
- Traducción automática
- Generación de texto
- Resumen automático
- Respuesta a preguntas
- Clasificación de texto

## 3. Introducción a BERT
BERT (Bidirectional Encoder Representations from Transformers) es un modelo preentrenado desarrollado por Google que entiende el contexto bidireccionalmente. Hugging Face permite usar variantes de BERT en múltiples idiomas, incluyendo español.

## 4. Análisis de Sentimientos con un Modelo en Español

```
from transformers import pipeline

# Modelo multilingüe para análisis de sentimientos
sentiment_analysis = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)

texto = "Me encantó esta película, fue una experiencia maravillosa."
resultado = sentiment_analysis(texto)
print(resultado)
```

## 5. Generación de Texto en Español

```
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "PlanTL-GOB-ES/gpt2-base-bne"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

prompt = "En un futuro cercano, la inteligencia artificial"
inputs = tokenizer(prompt, return_tensors="pt")

outputs = model.generate(
    **inputs,
    max_new_tokens=50,
    do_sample=True,
    top_k=50,
    top_p=0.95,
    temperature=0.9,
    num_return_sequences=1
)

generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(generated_text)
```

## 6. Resumen Automático de Textos en Español

```
from transformers import pipeline

# Modelo multilingüe para resumen
summarizer = pipeline("summarization", model="facebook/mbart-large-50-many-to-many-mmt")

texto_largo = '''
La inteligencia artificial (IA) está transformando muchos aspectos de la sociedad moderna.
Desde la medicina hasta la agricultura, la IA permite automatizar tareas complejas, generar
análisis predictivos, y mejorar la eficiencia de los procesos. A medida que esta tecnología
avanza, es importante considerar tanto sus beneficios como los retos éticos que conlleva.
'''

resumen = summarizer(texto_largo, max_length=60, min_length=20, do_sample=False)
print(resumen[0]['summary_text'])
```

## 7. Conclusión
Hugging Face ofrece una plataforma poderosa para usar modelos de NLP en múltiples tareas. Con pocas líneas de código, podemos aplicar modelos en español para análisis de sentimientos, generación de texto y resumen automático.
