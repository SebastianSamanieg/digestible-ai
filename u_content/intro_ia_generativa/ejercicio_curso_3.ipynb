```python
import re

# Textos de entrada
textos2 = [
    "¡INCREÍBLE!!! Compré el curso en https://miplataforma.com/curso?id=123  y me llegó el acceso al correo :)",
    "Mi usuario es @edward_latorre y el #cursoNLP estuvo brutal!!! 10/10 recomendado 🔥🔥🔥",
    "Actualización v2.0.1: ahora la app tarda 3 segundos en abrir (antes 12). Link: www.miapp.com/descarga",
    "Promoción válida hasta el 30/06/2026!!! Paga $250.000 o 250000 COP y recibe 2 meses gratis 😎",
    "El pedido #A-99321 llegó el día 2026-02-13, pero venía con 1 golpe en la esquina :(",
    "Soporte: escríbenos a soporte@empresa.com o visita https://empresa.com/ayuda. Gracias!!!",
    "Lo vi en TikTok @tutienda_oficial 😅 y decidí comprarlo... ERROR: no era lo que decía el video."
]

def normalizar_texto(texto):
    # 1. Convertir a minúsculas
    texto = texto.lower()

    # 2. Eliminar URLs (http, https, www)
    texto = re.sub(r'https?://\S+|www\.\S+', '', texto)

    # 3. Eliminar emojis y símbolos especiales
    texto = re.sub(r'[^\w\s]', '', texto)

    # 4. Eliminar números
    texto = re.sub(r'\d+', '', texto)

    # 5. Eliminar espacios múltiples
    texto = re.sub(r'\s+', ' ', texto)

    # 6. Eliminar espacios al inicio y al final
    texto = texto.strip()

    return texto

# Aplicar normalización a todos los textos
textos_normalizados = [normalizar_texto(texto) for texto in textos2]

# Mostrar resultados
print("TEXTOS ORIGINALES vs NORMALIZADOS\n")
print("=" * 80)
for i, (original, normalizado) in enumerate(zip(textos2, textos_normalizados), 1):
    print(f"\n{i}. ORIGINAL:")
    print(f"   {original}")
    print(f"\n   NORMALIZADO:")
    print(f"   {normalizado}")
    print("-" * 80)
```

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Download necessary data
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')

# Textos para el ejercicio 3
textos3 = [
    "Los estudiantes estudiaron bastante y estudian todos los días para el examen.",
    "La organización organizó actividades organizacionales para reorganizar el equipo.",
    "Compré, compras, comprando, compraría, compraremos: todo depende del presupuesto.",
    "El sistema falló, falla y fallará; mientras tanto seguimos intentando solucionarlo.",
    "Nos comunicamos, se comunicaron y se comunicarán por correo para confirmar la inscripción.",
    "La red neuronal aprende, aprendió y aprenderá patrones a partir de datos.",
    "Optimizar, optimización, optimizado: son palabras relacionadas pero no iguales."
]

# Inicializar stemmer y lemmatizer
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

def comparar_stemming_lemmatization(texto):
    # Tokenizar
    palabras = word_tokenize(texto.lower())

    # Aplicar stemming
    stems = [stemmer.stem(palabra) for palabra in palabras]

    # Aplicar lemmatization
    lemmas = [lemmatizer.lemmatize(palabra) for palabra in palabras]

    return palabras, stems, lemmas

print("=" * 100)
print("EJERCICIO 3: STEMMING vs LEMMATIZATION")
print("=" * 100)

for i, texto in enumerate(textos3, 1):
    print(f"\n{i}. TEXTO ORIGINAL:")
    print(f"   {texto}")

    palabras, stems, lemmas = comparar_stemming_lemmatization(texto)

    print(f"\n   PALABRAS TOKENIZADAS:")
    print(f"   {palabras}")

    print(f"\n   STEMMING (raíces):")
    print(f"   {stems}")

    print(f"\n   LEMMATIZATION (formas base):")
    print(f"   {lemmas}")

    print("\n   DIFERENCIAS NOTABLES:")
    for palabra, stem, lemma in zip(palabras, stems, lemmas):
        if stem != lemma and palabra.isalpha():
            print(f"   - '{palabra}' → Stem: '{stem}' | Lemma: '{lemma}'")

    print("-" * 100)
```

```python
import re

# Textos para el ejercicio 4
textos4 = [
    "John's email is john.doe@example.com and he bought 3 items for $29.99. Contact at SALES@company.com",

    "Hola soporte!!! Mi correo es juan.perez+promo@gmail.com y el alterno jperez@empresa.co. "
    "Mi celular: +57 312-555-0199. Ticket: TKT#9981. Gracias!!!",

    "Buenas, no puedo entrar :( Usuario: maria.lopez@outlook.com; tel: 300 123 4567. "
    "Pedido AB-1234 / ID: 000123. Urgente!!!",

    "Señores: favor confirmar la devolución. Contacto: compras@mi-tienda.com, cc: legal@mi-tienda.com. "
    "Teléfono fijo (601) 745-8899 ext. 123. Código caso: CASE-77XY.",

    "Mi correo: carlos_1990@yahoo.es y mi número 57-320-000-0000. "
    "Me llegó este mensaje raro: 'Error@500!! ###' y no sé qué hacer.",

    "Hola! escribir a: soporte@empresa.com; o a soporte2@empresa.com. "
    "Tel: +34 600-12-34-56. Ref: REF-2026-02-13. Gracias :)"
]

def extraer_datos(texto):
    # 1. Extraer números (enteros y decimales)
    numeros = re.findall(r'\d+\.?\d*', texto)

    # 2. Extraer correos electrónicos
    correos = re.findall(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', texto)

    # 3. Extraer palabras completamente en mayúsculas (mínimo 2 letras)
    mayusculas = re.findall(r'\b[A-Z]{2,}\b', texto)

    # 4. Extraer teléfonos (varios formatos)
    telefonos = re.findall(r'[\+\(]?\d{1,4}[\)\-\s]?\d{2,4}[\-\s]?\d{3,4}[\-\s]?\d{3,4}', texto)

    # 5. Extraer códigos/referencias (letras-números)
    codigos = re.findall(r'\b[A-Z]{2,}[\-#]?\d+[A-Z]*\b', texto)

    return {
        'numeros': numeros,
        'correos': correos,
        'mayusculas': mayusculas,
        'telefonos': telefonos,
        'codigos': codigos
    }

print("\n" + "=" * 100)
print("EJERCICIO 4: EXTRACCIÓN DE DATOS CON EXPRESIONES REGULARES")
print("=" * 100)

for i, texto in enumerate(textos4, 1):
    print(f"\n{i}. TEXTO:")
    print(f"   {texto}")

    datos = extraer_datos(texto)

    print(f"\n   📊 NÚMEROS ENCONTRADOS: {datos['numeros']}")
    print(f"   📧 CORREOS ELECTRÓNICOS: {datos['correos']}")
    print(f"   🔤 PALABRAS EN MAYÚSCULAS: {datos['mayusculas']}")
    print(f"   📱 TELÉFONOS: {datos['telefonos']}")
    print(f"   🏷️  CÓDIGOS/REFERENCIAS: {datos['codigos']}")

    print("-" * 100)

# BONUS: Función para análisis detallado
def analisis_completo(texto):
    print("\n" + "🔍 ANÁLISIS DETALLADO")
    print("=" * 80)

    # URLs
    urls = re.findall(r'https?://[^\s]+|www\.[^\s]+', texto)
    if urls:
        print(f"🌐 URLs: {urls}")

    # Menciones (@usuario)
    menciones = re.findall(r'@[\w]+', texto)
    if menciones:
        print(f"👤 Menciones: {menciones}")

    # Hashtags
    hashtags = re.findall(r'#[\w]+', texto)
    if hashtags:
        print(f"# Hashtags: {hashtags}")

    # Monedas
    monedas = re.findall(r'[\$€£]\s?\d+[,.]?\d*', texto)
    if monedas:
        print(f"💰 Monedas: {monedas}")

    # Fechas (varios formatos)
    fechas = re.findall(r'\d{1,2}[/-]\d{1,2}[/-]\d{2,4}|\d{4}[/-]\d{2}[/-]\d{2}', texto)
    if fechas:
        print(f"📅 Fechas: {fechas}")

# Ejemplo de uso del análisis completo
print("\n\n" + "=" * 100)
print("BONUS: ANÁLISIS COMPLETO DE ALGUNOS TEXTOS")
print("=" * 100)

for texto in textos4[:3]:  # Analizar los primeros 3 textos
    print(f"\nTexto: {texto[:60]}...")
    analisis_completo(texto)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Sample text data
documents = [
    "Barack Obama was the 44th President of the United States",
    "The President lives in the White House",
    "The United States has a strong economy"
]

# Initialize the CountVectorizer
vectorizer = CountVectorizer()

# Fit the model and transform the documents into a Bag of Words
bow_matrix = vectorizer.fit_transform(documents)

# Get the feature names (unique words in the corpus)
feature_names = vectorizer.get_feature_names_out()

# Convert the Bag of Words matrix into an array
bow_array = bow_matrix.toarray()

# Display the Bag of Words
print("Feature Names (Words):", feature_names)
print("\nBag of Words Representation:")
print(bow_array)
```
