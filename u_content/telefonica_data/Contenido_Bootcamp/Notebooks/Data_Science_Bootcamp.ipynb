### Tipos de Variables

## Tipos de Variables en Análisis de Datos

En el análisis de datos, las **variables** son las características que describen a los elementos de un conjunto de datos. Estas variables pueden ser de diferentes tipos, y su clasificación es fundamental para elegir las técnicas estadísticas adecuadas para analizar la información.

### Clasificación de las Variables

Las variables se pueden clasificar en dos grandes grupos:

### 1. Variables Cuantitativas
Representan cantidades numéricas y pueden ser medidas.

- **Discretas**: Toman valores enteros y representan cantidades que se pueden contar.
  **Ejemplos**: número de hijos, cantidad de carros en un estacionamiento, número de llamadas recibidas.

- **Continuas**: Pueden tomar cualquier valor dentro de un intervalo y representan medidas.
  **Ejemplos**: altura, peso, temperatura, tiempo.

### 2. Variables Cualitativas (Categóricas)
Expresan cualidades o características que no pueden ser medidas numéricamente.

- **Nominales**: No existe un orden natural entre las categorías.
  **Ejemplos**: color de cabello, género, país de origen.

- **Ordinales**: Existe un orden natural entre las categorías, pero no se pueden medir las distancias entre ellas.
  **Ejemplos**: nivel de estudios (primaria, secundaria, universitaria), escala de Likert (muy de acuerdo, de acuerdo, ni de acuerdo ni en desacuerdo, en desacuerdo, muy en desacuerdo).
