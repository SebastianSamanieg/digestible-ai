### ¿Qué es CRISP-DM?

CRISP-DM es la metodología estándar más utilizada en proyectos de minería de datos y ciencia de datos. Define un proceso **iterativo** de 6 fases que guía desde la comprensión del negocio hasta el despliegue de soluciones. Su fortaleza radica en ser **agnóstica a la industria y a las herramientas**, lo que la hace aplicable en cualquier contexto.
