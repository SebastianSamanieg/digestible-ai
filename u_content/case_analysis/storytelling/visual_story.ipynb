## 📊 1. Análisis por Regiones

- `Municipio` 🏘️

```python
# Configuración general
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

# 1️⃣ Distribución por Regional

# --- 3️⃣ Distribución por Regional ---
plt.figure()
sns.countplot(
    data=df_story,
    x='Regional',
    hue='Regional',  # ✅ misma variable como hue
    order=df_story['Regional'].value_counts().index,
    palette='flare',
    legend=False
)
plt.title("Distribución por Regional")
plt.xlabel("Regional")
plt.ylabel("Cantidad de Personas")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2️⃣ Distribución por Municipio (Top 10)
mun_counts = df_story['Municipio'].value_counts().head(10)

plt.figure()
mun_counts.plot(kind='bar', color='mediumseagreen', edgecolor='black')
plt.title('Top 10 Municipios con más Participantes', fontsize=14)
plt.xlabel('Municipio')
plt.ylabel('Número de Participantes')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# 3️⃣ Distribución de Edades
plt.figure()
df_story['Edad'].plot(kind='hist', bins=20, color='salmon', edgecolor='black')
plt.title('Distribución de Edad de las Participantes', fontsize=14)
plt.xlabel('Edad')
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()

# 4️⃣ Promedio de edad por regional
edad_prom = df_story.groupby('Regional')['Edad'].mean().sort_values(ascending=False)

plt.figure()
edad_prom.plot(kind='bar', color='plum', edgecolor='black')
plt.title('Promedio de Edad por Regional', fontsize=14)
plt.xlabel('Regional')
plt.ylabel('Edad promedio')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# 5️⃣ Estadísticas descriptivas de edad
df_story['Edad'].describe()
```

## 🌍 2. Análisis de Distribución en el País

---

En esta sección se examina cómo se distribuyen las personas participantes a nivel nacional, considerando variables clave que permiten comprender la **composición demográfica, educativa y laboral** de la población atendida en los programas.

### 🎯 Objetivo
Identificar **patrones de concentración y diversidad** en las principales dimensiones socioeconómicas, como **rango de edad**, **nivel educativo**, **región** y **ocupación**.

---

### 👥 Distribución por Rango de Edad
La variable `rango_edad` agrupa a las personas según intervalos etarios definidos, lo que permite analizar la participación relativa de **jóvenes, adultos y adultos mayores**.
Este análisis facilita identificar si la estrategia del programa logra un alcance equilibrado entre grupos generacionales o si existen **cohortes subrepresentadas** que podrían requerir acciones focalizadas.

---

### 🎓 Nivel Educativo
El nivel educativo refleja el **grado de formación alcanzado** por las personas participantes.
Esta información es esencial para detectar **brechas de acceso a la educación** y orientar los contenidos formativos o acompañamientos según las capacidades y contextos del público objetivo.

---

### 🗺️ Distribución por Regional
El análisis por la variable `Regional` evidencia la **cobertura territorial** del programa en las distintas zonas del país.
A partir de esta distribución, se pueden identificar las **regiones con mayor participación** y aquellas donde podría fortalecerse la presencia institucional para lograr una mayor equidad territorial.

---

### 💼 Ocupación
La variable `Ocupacion` muestra la diversidad de actividades laborales desempeñadas por las personas beneficiarias.
El análisis se centra en las **ocupaciones más frecuentes**, permitiendo reconocer los perfiles predominantes e identificar **áreas con potencial de desarrollo o capacitación complementaria**.

---

En conjunto, estos análisis permiten construir una visión integral de la población participante, combinando **perspectivas demográficas, educativas, territoriales y laborales**, fundamentales para la toma de decisiones y la comunicación estratégica basada en evidencia.

---


```python
# ==============================
# 2️⃣ Análisis de Distribución en el País
# ==============================

import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Configuración visual y supresión de warnings irrelevantes
warnings.filterwarnings("ignore", category=UserWarning, module="matplotlib")
warnings.filterwarnings("ignore", category=FutureWarning, module="seaborn")

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 12

# --- 1️⃣ Distribución por Rango de Edad ---
plt.figure()
sns.countplot(
    data=df_story,
    x='rango_edad',
    hue='rango_edad',  # ✅ compatibilidad futura
    order=df_story['rango_edad'].value_counts().index,
    palette='crest',
    legend=False
)
plt.title("Distribución por Rango de Edad")
plt.xlabel("Rango de Edad")
plt.ylabel("Número de Personas")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# --- 2️⃣ Nivel Educativo ---
plt.figure()
sns.countplot(
    data=df_story,
    y='Educacion',
    hue='Educacion',  # ✅ misma variable como hue
    order=df_story['Educacion'].value_counts().index,
    palette='viridis',
    legend=False
)
plt.title("Distribución por Nivel Educativo")
plt.xlabel("Número de Personas")
plt.ylabel("Nivel Educativo")
plt.tight_layout()
plt.show()

# --- 4️⃣ Distribución por Ocupación ---
plt.figure()
top_ocupaciones = df_story['Ocupacion'].value_counts().nlargest(10).reset_index()
top_ocupaciones.columns = ['Ocupacion', 'Cantidad']

sns.barplot(
    data=top_ocupaciones,
    x='Cantidad',
    y='Ocupacion',
    hue='Ocupacion',  # ✅ se agrega hue para evitar FutureWarning
    palette='mako',
    legend=False
)
plt.title("Principales Ocupaciones (Top 10)")
plt.xlabel("Cantidad de Personas")
plt.ylabel("Ocupacion")
plt.tight_layout()
plt.show()
```

## 🌈 4. Análisis de Diversidad e Inclusión

---

En este apartado se examina la **representación de distintos grupos poblacionales** dentro del programa, con el fin de identificar niveles de **inclusión y participación equitativa**.

Los indicadores observados permiten evaluar la presencia de grupos históricamente subrepresentados o vulnerables:

- **Etnia:** distribución de participantes por pertenencia a comunidades étnicas.
- **Comunidad LGBTIQ+:** participación de personas que se reconocen dentro de esta comunidad.
- **Discapacidad:** proporción de participantes que reportan alguna discapacidad.
- **Regional:** enfoque territorial para observar si existen diferencias significativas entre regiones.

Este análisis contribuye a entender cómo las políticas públicas o los programas están alcanzando a poblaciones diversas, fortaleciendo la **equidad territorial y social**.

---


```python
# ==============================
# 4️⃣ Análisis de Diversidad e Inclusión (optimizado)
# ==============================

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 12

# --- 1️⃣ Distribución por Etnia ---
df_etnia = df_story[~df_story['Etnia'].str.lower().isin(['no', 'ninguna', 'nan'])]
plt.figure()
ax = sns.countplot(
    data=df_etnia,
    y='Etnia',
    order=df_etnia['Etnia'].value_counts().index,
    palette='crest',
    hue='Etnia',
    legend=False
)
plt.title("Distribución por Etnia (excluyendo 'No' o 'Ninguna')")
plt.xlabel("Número de Personas")
plt.ylabel("Etnia")

# Añadir etiquetas con valores
for p in ax.patches:
    width = p.get_width()
    ax.text(width + 50, p.get_y() + p.get_height() / 2,
            f'{int(width):,}', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# --- 2️⃣ Pertenencia a Comunidad LGBTIQ+ ---
plt.figure()
ax = sns.countplot(
    data=df_story,
    x='Comunidad lgbtiq',
    hue='Comunidad lgbtiq',
    palette='Set2',
    legend=False
)
plt.title("Participación en la Comunidad LGBTIQ+")
plt.xlabel("¿Pertenece a la Comunidad LGBTIQ+?")
plt.ylabel("Número de Personas")

# Etiquetas numéricas
for p in ax.patches:
    height = p.get_height()
    ax.text(p.get_x() + p.get_width()/2, height + 100,
            f'{int(height):,}', ha='center', fontsize=10)

plt.tight_layout()
plt.show()

# --- 3️⃣ Condición de Discapacidad ---
df_discap = df_story[~df_story['Discapacidad'].str.lower().isin(['no', 'ninguna', 'nan'])]
plt.figure()
ax = sns.countplot(
    data=df_discap,
    x='Discapacidad',
    hue='Discapacidad',
    palette='coolwarm',
    legend=False
)
plt.title("Distribución de Personas con Discapacidad (excluyendo 'No')")
plt.xlabel("Tipo de Discapacidad")
plt.ylabel("Número de Personas")

# Etiquetas numéricas
for p in ax.patches:
    height = p.get_height()
    ax.text(p.get_x() + p.get_width()/2, height + 50,
            f'{int(height):,}', ha='center', fontsize=10)

plt.tight_layout()
plt.show()

# --- 4️⃣ Distribución Regional de Comunidad LGBTIQ+ ---
plt.figure()
ax = sns.countplot(
    data=df_story,
    x='Regional',
    hue='Comunidad lgbtiq',
    palette='mako'
)
plt.title("Distribución Regional de la Comunidad LGBTIQ+")
plt.xlabel("Regional")
plt.ylabel("Cantidad de Personas")
plt.xticks(rotation=45)
plt.legend(title="Comunidad LGBTIQ+")

# Etiquetas numéricas en barras
for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', fontsize=9)

plt.tight_layout()
plt.show()
```

## 🎓 8. Análisis de Finalización y Tasa de Éxito

---

En esta sección se evalúa el **comportamiento del proceso formativo** en términos de:

- **Inscripción:** Participación de mujeres en cada curso.
- **Finalización:** Personas que completaron los cursos.
- **Certificación:** Mujeres que obtuvieron un certificado tras completar el curso.

El análisis abarca los dos cursos principales:
- *Curso N°1: Marketing Digital*
- *Curso N°2: Aplicaciones en Línea*

Se calculan métricas clave para comprender la eficiencia del programa:

| Métrica | Descripción |
|----------|--------------|
| **Tasa de Inscripción** | Porcentaje de personas inscritas respecto al total del universo. |
| **Tasa de Finalización** | Porcentaje de personas que completaron el curso respecto a las inscritas. |
| **Tasa de Certificación** | Porcentaje de certificadas respecto a las que finalizaron. |

Estas métricas ayudan a **evaluar la eficacia, el compromiso y la retención** del programa, permitiendo identificar áreas de mejora en los procesos formativos y de acompañamiento.

---


```python
# ==============================
# 8️⃣ Análisis de Finalización (versión corregida y validada)
# ==============================

# --- 1️⃣ Normalización y limpieza de datos ---
# Convertimos certificados a valores binarios (0/1)
df_story['Certificados Curso No 1 - Marketing Digital'] = pd.to_numeric(
    df_story['Certificados Curso No 1 - Marketing Digital'], errors='coerce'
).fillna(0).astype(int)

df_story['Certificados Curso No 2 - Aplicaciones en linea'] = pd.to_numeric(
    df_story['Certificados Curso No 2 - Aplicaciones en linea'], errors='coerce'
).fillna(0).astype(int)

# Normalizamos texto en los detalles de cursos
df_story['Detalle Curso N°1'] = df_story['Detalle Curso N°1'].astype(str).str.strip().str.lower()
df_story['Detalle Curso N°2'] = df_story['Detalle Curso N°2'].astype(str).str.strip().str.lower()

# --- 2️⃣ Filtramos los inscritos reales (excluyendo "no aplica") ---
inscritos_c1 = df_story[df_story['Detalle Curso N°1'] != 'no aplica']
inscritos_c2 = df_story[df_story['Detalle Curso N°2'] != 'no aplica']

# --- 3️⃣ Finalizados (solo entre los inscritos) ---
finalizados_c1 = inscritos_c1['Certificados Curso No 1 - Marketing Digital'].sum()
finalizados_c2 = inscritos_c2['Certificados Curso No 2 - Aplicaciones en linea'].sum()

# --- 4️⃣ Cálculo de tasas ---
total_participantes = len(df_story)

# Tasa de inscripción (proporción de personas inscritas sobre el total)
tasa_inscripcion_c1 = (len(inscritos_c1) / total_participantes) * 100 if total_participantes > 0 else 0
tasa_inscripcion_c2 = (len(inscritos_c2) / total_participantes) * 100 if total_participantes > 0 else 0

# Tasa de finalización (proporción de finalizados sobre inscritos)
tasa_finalizacion_c1 = (finalizados_c1 / len(inscritos_c1)) * 100 if len(inscritos_c1) > 0 else 0
tasa_finalizacion_c2 = (finalizados_c2 / len(inscritos_c2)) * 100 if len(inscritos_c2) > 0 else 0

# --- 5️⃣ Tasa de deserción (solo entre inscritos) ---
# Deserción = porcentaje de inscritos que no finalizaron
tasa_desercion_c1 = 100 - tasa_finalizacion_c1 if len(inscritos_c1) > 0 else 0
tasa_desercion_c2 = 100 - tasa_finalizacion_c2 if len(inscritos_c2) > 0 else 0

# --- 6️⃣ Construcción del DataFrame para visualización ---
data = pd.DataFrame({
    'Curso': ['Marketing Digital', 'Marketing Digital', 'Marketing Digital',
              'Aplicaciones en Línea', 'Aplicaciones en Línea', 'Aplicaciones en Línea'],
    'Etapa': ['Tasa de Inscripción', 'Tasa de Finalización', 'Tasa de Deserción',
              'Tasa de Inscripción', 'Tasa de Finalización', 'Tasa de Deserción'],
    'Porcentaje': [tasa_inscripcion_c1, tasa_finalizacion_c1, tasa_desercion_c1,
                   tasa_inscripcion_c2, tasa_finalizacion_c2, tasa_desercion_c2]
})

# --- 7️⃣ Visualización ---
plt.figure(figsize=(10, 6))
sns.barplot(
    data=data,
    x='Etapa',
    y='Porcentaje',
    hue='Curso',
    palette='crest',
    legend=True
)

plt.title("Tasas de Participación, Finalización y Deserción por Curso")
plt.xlabel("Etapa del Proceso")
plt.ylabel("Porcentaje (%)")
plt.ylim(0, 110)

# Etiquetas de porcentaje sobre las barras
for p in plt.gca().patches:
    plt.gca().annotate(
        f"{p.get_height():.1f}%",
        (p.get_x() + p.get_width() / 2., p.get_height() + 1),
        ha='center', va='bottom', fontsize=9
    )

plt.tight_layout()
plt.show()

# --- 8️⃣ Resumen impreso ---
print("📊 RESUMEN DE TASAS:")
print(f"Marketing Digital → Inscripción: {tasa_inscripcion_c1:.1f}%, Finalización: {tasa_finalizacion_c1:.1f}%, Deserción: {tasa_desercion_c1:.1f}%")
print(f"Aplicaciones en Línea → Inscripción: {tasa_inscripcion_c2:.1f}%, Finalización: {tasa_finalizacion_c2:.1f}%, Deserción: {tasa_desercion_c2:.1f}%")
```
