## ⚙️ 3.1 Normalización de Datos

---

En esta etapa se lleva a cabo la **homogeneización de valores dentro del dataset**, con el objetivo de garantizar la coherencia y consistencia de la información.
Este proceso incluye la **limpieza de texto**, **unificación de categorías equivalentes** y la **corrección de formatos** en variables clave.

**Principales acciones realizadas:**
- Eliminación de espacios innecesarios, mayúsculas inconsistentes y tildes.
- Estandarización de categorías en variables como `Ocupacion`, `Genero` o `Departamento`.
- Sustitución de valores duplicados o ambiguos por etiquetas unificadas.
- Conversión de datos textuales a formatos uniformes (por ejemplo, minúsculas).

El resultado de esta etapa es un conjunto de datos **más limpio, consistente y apto para análisis estadísticos o modelado posterior**.

---


```python
# ───────────────────────────────────────────────────────────────────────────────
# Normalización de la columna "Ocupacion"
# ───────────────────────────────────────────────────────────────────────────────

print("\n" + "="*80)
print("🔹 ETAPA: NORMALIZACIÓN DE LA COLUMNA 'Ocupacion'")
print("="*80 + "\n")

print("📊 Valores originales (antes de la normalización):")
print("-"*80)
print(df["Ocupacion"].value_counts(dropna=False))
print("-"*80 + "\n")

df_clean["Ocupacion"] = (
    df_clean["Ocupacion"]
        .astype(str)
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
        .str.lower()
        .replace({
            "servidor público": "servidora pública",
            "trabajador independiente": "trabajadora independiente"
        })
        .map({
            "no aplica": "No aplica / No registra",
            "no registra": "No aplica / No registra",
            "estudiante": "Estudiante",
            "independiente": "Independiente",
            "trabajadora independiente": "Independiente",
            "empleada": "Empleada",
            "docente": "Empleada",
            "servidora pública": "Empleada",
            "desempleada": "Desempleada",
            "pensionada": "Pensionada",
            "hogar": "Ama de casa / No remunerada",
            "en actividades no remuneradas": "Ama de casa / No remunerada"
        })
)

print("✅ Valores normalizados (después de aplicar transformaciones):")
print("-"*80)
print(df_clean["Ocupacion"].value_counts(dropna=False))
print("-"*80)

print(f"\n🔎 Total de registros procesados: {len(df_clean):,}")
print(f"🚫 Registros sin valor asignado (NaN): {df_clean['Ocupacion'].isna().sum():,}")
print("="*80 + "\n")
```

## 🧮 3.2 Manejo y Creación de Nuevas Columnas

---

En esta sección se abordan los procesos de **tratamiento, derivación y enriquecimiento de columnas** del dataset.
Se crean nuevas variables a partir de la información existente, con el fin de facilitar el análisis y segmentación.

**Principales acciones realizadas:**
- Generación de nuevas variables derivadas de combinaciones o transformaciones lógicas.
- Verificación de la correcta asignación de valores en las nuevas columnas.

Estas transformaciones permiten **ampliar las posibilidades de análisis**, mejorando la comprensión de los datos y su comportamiento por grupos o categorías.

```python
# ───────────────────────────────────────────────────────────────────────────────
# Creación de la columna 'rango_edad'
# ───────────────────────────────────────────────────────────────────────────────

print("\n" + "="*80)
print("🧮 ETAPA: CREACIÓN DE LA COLUMNA 'rango_edad'")
print("="*80 + "\n")

# Definir los rangos de edad y sus etiquetas
bins = [0, 14, 19, 29, 44, 59, float('inf')]
labels = ['0-14 años', '15-19 años', '20-29 años', '30-44 años', '45-59 años', '60+ años']

# Crear la nueva columna categórica
df_clean['rango_edad'] = pd.cut(df_clean['Edad'], bins=bins, labels=labels, right=True)

# ───────────────────────────────────────────────────────────────────────────────
# Validación de la creación del rango de edad
# ───────────────────────────────────────────────────────────────────────────────

print("📊 Distribución de registros por rango de edad:")
print("-"*80)
print(df_clean['rango_edad'].value_counts(dropna=False).sort_index())
print("-"*80)

# Mostrar ejemplos de asignación
print("\n👀 Ejemplo de registros con su rango de edad asignado:")
print("-"*80)
print(df_clean[['Edad', 'rango_edad']].head(10).to_string(index=False))
print("-"*80)

# Estadísticas complementarias
print(f"\n🔎 Total de registros procesados: {len(df_clean):,}")
print(f"🚫 Registros sin rango asignado (NaN): {df_clean['rango_edad'].isna().sum():,}")
print("="*80 + "\n")
```

```python
# ───────────────────────────────────────────────────────────────────────────────
# Creación de la columna "Regional" a partir del código DANE del departamento
# ───────────────────────────────────────────────────────────────────────────────

print("\n" + "="*80)
print("🗺️ ETAPA: CREACIÓN DE LA COLUMNA 'Regional'")
print("="*80 + "\n")

# Diccionario de asignación por código DANE
regional_map = {
    # Región Caribe
    8:  "Caribe",
    13: "Caribe",
    20: "Caribe",
    23: "Caribe",
    44: "Caribe",
    47: "Caribe",
    70: "Caribe",
    88: "Caribe",

    # Región Pacífica
    27: "Pacífica",
    19: "Pacífica",
    52: "Pacífica",
    76: "Pacífica",

    # Región Andina
    5:  "Andina",
    11: "Andina",
    15: "Andina",
    17: "Andina",
    25: "Andina",
    41: "Andina",
    54: "Andina",
    63: "Andina",
    66: "Andina",
    68: "Andina",
    73: "Andina",

    # Región Orinoquía
    81: "Orinoquía",
    85: "Orinoquía",
    50: "Orinoquía",
    99: "Orinoquía",

    # Región Amazonía
    91: "Amazonía",
    18: "Amazonía",
    94: "Amazonía",
    95: "Amazonía",
    86: "Amazonía",
    97: "Amazonía",

    # Sin clasificar
    None: "Sin Clasificar"
}

# Crear nueva columna "Regional"
df_clean["Regional"] = df_clean["Departamento Dane"].map(regional_map)

# ───────────────────────────────────────────────────────────────────────────────
# Validación de la nueva columna
# ───────────────────────────────────────────────────────────────────────────────

print("📊 Distribución de registros por región:")
print("-"*80)
print(df_clean["Regional"].value_counts(dropna=False))
print("-"*80)

print("\n👀 Ejemplo de asignación de regiones:")
print(df_clean[["Departamento Dane", "Regional"]].head(10).to_string(index=False))
print("="*80 + "\n")
```

```python
# ───────────────────────────────────────────────────────────────────────────────
# Creación de las columnas "Detalle Curso N°1" y "Detalle Curso N°2"
# ───────────────────────────────────────────────────────────────────────────────

print("\n" + "="*80)
print("🎓 ETAPA: CREACIÓN DE COLUMNAS DE DETALLE DE CURSOS")
print("="*80 + "\n")

# Limpieza básica previa del texto
df_clean["Cursos"] = (
    df_clean["Cursos"]
        .astype(str)
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
)

# Definir funciones para identificar los cursos
def obtener_curso_1(curso_str):
    curso_str = str(curso_str).strip().lower()
    if "mujeres creadoras de contenido digital" in curso_str:
        return "Marketing Digital (Mujeres creadoras de contenido digital)"
    elif curso_str in ["no aplica", "ninguno", "nan", ""]:
        return "No aplica"
    else:
        return "No aplica"

def obtener_curso_2(curso_str):
    curso_str = str(curso_str).strip().lower()
    if "mujeres líderes de la transformación digital" in curso_str:
        return "Aplicaciones en línea (Mujeres líderes de la transformación digital)"
    elif curso_str in ["no aplica", "ninguno", "nan", ""]:
        return "No aplica"
    else:
        return "No aplica"

# Aplicar funciones
df_clean["Detalle Curso N°1"] = df_clean["Cursos"].apply(obtener_curso_1)
df_clean["Detalle Curso N°2"] = df_clean["Cursos"].apply(obtener_curso_2)

# ───────────────────────────────────────────────────────────────────────────────
# Validación de las nuevas columnas
# ───────────────────────────────────────────────────────────────────────────────

print("📊 Distribución de registros por tipo de curso:")
print("-"*80)
print("Curso N°1:")
print(df_clean["Detalle Curso N°1"].value_counts(dropna=False))
print("\nCurso N°2:")
print(df_clean["Detalle Curso N°2"].value_counts(dropna=False))
print("-"*80)

print("\n👀 Ejemplo de asignación de cursos:")
print(df_clean[["Cursos", "Detalle Curso N°1", "Detalle Curso N°2"]].head(10).to_string(index=False))
print("="*80 + "\n")
```

## 🧹 3.3 Eliminación de Datos

---

Durante el proceso de depuración, se identificaron **registros y columnas que no aportan valor analítico** o que pueden **introducir ruido o sesgos** en el análisis posterior.
La eliminación de estos elementos se realiza con el fin de **mejorar la calidad, precisión y rendimiento** del dataset.

**Principales motivos para la eliminación:**
- Registros con **valores nulos o inconsistentes** en variables críticas.
- Columnas redundantes o derivadas de otras variables ya consideradas.
- Datos fuera del alcance del estudio o que no cumplen con los criterios definidos.
- Estandarización del dataset para facilitar su interpretación y visualización.

**Justificación:**
Eliminar información innecesaria permite obtener un conjunto de datos **más limpio, relevante y manejable**, optimizando tanto el procesamiento como la interpretación de los resultados.

---


```python
# ───────────────────────────────────────────────────────────────────────────────
# ELIMINACIÓN DE LA COLUMNA "nombre de red"
# ───────────────────────────────────────────────────────────────────────────────

print("="*80)
print("⚠️  EVALUACIÓN DE CALIDAD: ELIMINACIÓN DE COLUMNA 'nombre de red'")
print("="*80 + "\n")

nota_eliminacion = """
📋 NOTA TÉCNICA - DECISIÓN DE ELIMINACIÓN:

Motivo de Eliminación:
La columna 'nombre de red' fue eliminada del dataset debido a que NO GENERA VALOR
ANALÍTICO por las siguientes razones:

1. LLENADO MANUAL - Alta Variabilidad:
   • Los datos fueron ingresados de forma manual y descentralizada
   • Falta de estándares de validación en la entrada de datos
   • Inconsistencias significativas en nomenclatura y formato

2. RUIDO Y DUPLICACIÓN:
   • La columna "Red mujeres" nos dice si pertenece a una red
   • Múltiples variaciones del mismo nombre (ej: "tic mujer", "TIC mujer",
     "Por TIC mujer", "ticmujer")
   • Valores de ruido frecuentes: "no", "no registra", "jjjj", emails incompletos
   • Después de realizado el EDAA este indica alta dispersión

3. FALTA DE CONCORDANCIA:
   • La información no concuerda internamente (repeticiones sin patrón)
   • Imposible establecer una relación confiable con otras variables
   • No hay forma de validar la información contra fuentes externas

4. BAJO POTENCIAL PREDICTIVO:
   • La columna no correlaciona significativamente con el objetivo analítico
   • Su precisión post-normalización es insuficiente para análisis
   • Genera más complejidad que información útil
"""

# Eliminar la columna
df_clean = df_clean.drop(columns=["nombre de red"])

print("="*80)
print("✅ Columna 'nombre de red' eliminada del dataset")
print(f"📊 Dimensión actual del dataset: {df_clean.shape[0]:,} filas × {df_clean.shape[1]} columnas")
print("="*80 + "\n")
```
